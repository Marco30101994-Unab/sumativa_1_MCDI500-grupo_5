## 2 Contextualización del Proyecto

El proyecto aborda el análisis de los permisos activos de uso de calles en la ciudad de San Francisco, utilizando el conjunto de datos público disponible a través de DataSF. El conjunto de datos contiene información asociada a permisos de uso del espacio
público, el cual permite analizar características relacionadas con el tipo de permiso, su ubicación territorial y otros atributos disponibles. El análisis busca identificar patrones y factores asociados con la duración de estos permisos, mediante técnicas de exploración, procesamiento y análisis de datos.
