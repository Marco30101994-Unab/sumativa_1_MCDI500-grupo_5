## 3 Diagnóstico Inicial de Completitud de Datos
Se revisará la presencia de valores faltantes en el conjunto de datos original. En este punto, se descibirá la calidad inicial de la información, sin modificar ni eliminar registros durante la Fase 1.

```python
from pathlib import Path
import pandas as pd

ruta_proyecto = Path.cwd().parents[1]

ruta_datos = ruta_proyecto / "data" / "raw" / "Active_Street-Use_Permits_20260909.csv"

df = pd.read_csv(ruta_datos)

df.head()
```

```python
df.shape
```

```python
proporcion_nulos = df.isna().mean()
proporcion_nulos
```

```python
porcentaje_nulos = proporcion_nulos * 100
porcentaje_nulos_ordenado = porcentaje_nulos.sort_values(ascending=False)
porcentaje_nulos_ordenado
```

```python
tabla_nulos = porcentaje_nulos_ordenado.reset_index()
tabla_nulos.columns = ["Variable", "Porcentaje de nulos"]
tabla_nulos["Porcentaje de nulos"] = tabla_nulos["Porcentaje de nulos"].round(2)
tabla_nulos = tabla_nulos[
    tabla_nulos["Porcentaje de nulos"]>0
    ]
tabla_nulos
```
